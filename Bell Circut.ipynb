```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram, plot_bloch_multivector
import matplotlib.pyplot as plt
```

```python

# Step 1: Create 2-qubit circuit (starts as |00>)
qc = QuantumCircuit(2)
```

```python
# Step 2: Hadamard on control (qubit 0) → superposition |+> = (|0> + |1>)/√2
qc.h(0)
```

```python
# Step 3: CNOT (control 0, target 1) → entanglement (|00> + |11>)/√2
qc.cx(0, 1)
```

```python
print("Circuit diagram:")
print(qc.draw())
```

```python
# Visualize state before measurement (Bloch spheres show entanglement)
print("\nState visualization:")
plot_bloch_multivector(qc).show()  # Or .savefig('bloch.png')
```

```python
# Add measurements
qc.measure_all()
```

```python
# Simulate 1024 runs
simulator = AerSimulator()
```

```python
result = simulator.run(qc, shots=1024).result()
```

```python
counts = result.get_counts(qc)
```

```python
print("\nMeasurement results (roughly 50% '00', 50% '11'):")
print(counts)
```

```python
# Histogram plot
plot_histogram(counts).show()  # Shows the Bell state probabilities
```
